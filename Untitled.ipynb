**Spearman korelacija**

```python
numeric_cols = data.drop(columns=['is_safe']).columns

# Spearman korelacija između atributa i cilja
spearman_corr = data[numeric_cols].corrwith(data['is_safe'], method='spearman')
print("Spearman korelacija sa ciljem:")
print(spearman_corr.sort_values(ascending=False))

# Heatmap Spearman korelacija između atributa
plt.figure(figsize=(30,15))
sns.heatmap(data[numeric_cols].corr(method='spearman'), annot=True, cmap=cmap)
plt.title("Spearman korelacija između atributa")
plt.show()


X = data[numeric_cols]
y = data['is_safe']

# MI između svakog atributa i cilja
mi_scores = mutual_info_classif(X, y, discrete_features='auto', random_state=42)
mi_series = pd.Series(mi_scores, index=numeric_cols)
print("Mutual Information sa ciljem:")
print(mi_series.sort_values(ascending=False))

colors = sns.light_palette("lightcoral", n_colors=len(mi_series), reverse=False)


# Visualizacija MI
plt.figure(figsize=(10,6))
#colors = sns.light_palette("purple", n_colors=len(mi_series))
mi_series.sort_values().plot(kind='barh', color=colors)
plt.title("Mutual Information: atributi vs cilj")
plt.show()
```

```python
def series_analysis(column: pd.Series, column_name):
    # Pretvaranje u numerički tip i ignorisanje NaN vrednosti
    column = pd.to_numeric(column, errors='coerce').dropna()
    
    res_dict = dict()
    res_dict["name"] = column_name
    sorted_column = column.sort_values()

    res_dict["mean"] = column.mean()
    res_dict["median"] = column.median()
    res_dict["mode"] = column.mode().values.tolist()
    res_dict["geometric_mean"] = column.prod() ** (1.0 / column.count()) if column.count() > 0 else np.nan
    res_dict["harmonic_mean"] = column.count() / sum(1.0 / (item + 1e-10) for item in column.values) if column.count() > 0 else np.nan
    res_dict["max"] = float(column.max())
    res_dict["min"] = float(column.min())
    res_dict["range"] = float(column.max() - column.min())
    res_dict["std"] = column.std(ddof=1)
    res_dict["variance"] = column.var(ddof=1)
    res_dict["skewness"] = skew(column)
    res_dict["kurtosis"] = kurtosis(column)
    res_dict["iqr"] = iqr(sorted_column.values)
    res_dict["sorted"] = sorted_column.values.tolist()

    return res_dict

# Funkcija za deskriptivnu analizu celog dataset-a
def descriptive_analysis(dataset: pd.DataFrame):
    # Samo numeričke kolone
    numeric_data = dataset.select_dtypes(include=[np.number]).copy()
    
    description = numeric_data.describe()
    covariance_matrix = np.cov(numeric_data.values, rowvar=False, ddof=1)

    column_stats = []
    column_dict = {}
    for col in numeric_data.columns:
        temp = series_analysis(numeric_data[col], col)
        column_stats.append(temp)
        temp2 = copy.deepcopy(temp)
        del temp2["name"]
        column_dict[col] = temp2

    return description, covariance_matrix, column_stats, column_dict

# Primena funkcije
description, cov_matrix, column_stats, column_dict = descriptive_analysis(data)

# Prikaz rezultata
pd.set_option('display.max_columns', None)
print("Deskriptivna statistika:")
display(description)


print("\nDetaljna analiza po kolonama:")
display(pd.DataFrame(column_dict))

pd.reset_option('display.max_columns')
```
